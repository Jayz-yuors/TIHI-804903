```python
import numpy as np
import torch
import torch.nn as nn

# Simulate 1000 training samples and 200 validation samples
np.random.seed(42)
X_train_np = np.random.rand(1000, 5).astype(np.float32)
y_train_np = np.random.randint(0, 4, size=(1000,))

X_val_np = np.random.rand(200, 5).astype(np.float32)
y_val_np = np.random.randint(0, 4, size=(200,))

# Push to GPU
dev = "cuda" if torch.cuda.is_available() else "cpu"
X_train = torch.tensor(X_train_np, device=dev)
y_train = torch.tensor(y_train_np, dtype=torch.long, device=dev)
X_val = torch.tensor(X_val_np, device=dev)
y_val = torch.tensor(y_val_np, dtype=torch.long, device=dev)

print("Training and Validation tensors ready on GPU!")
```

```python
net = nn.Sequential(
    nn.Linear(5, 32),
    nn.BatchNorm1d(32), # Normalizes the 32 outputs
    nn.ReLU(),
    nn.Dropout(0.3),    # Randomly drops 30% of neurons during training

    nn.Linear(32, 16),
    nn.BatchNorm1d(16),
    nn.ReLU(),
    nn.Dropout(0.3),

    nn.Linear(16, 4)
).to(dev)

loss_fn = nn.CrossEntropyLoss()
# 1e-3 is scientific notation for 0.001
optimizer = torch.optim.Adam(net.parameters(), lr=1e-3)

print("Advanced MLP built with BatchNorm and Dropout!")
```

```python
print("Starting Day 7 Training (50 Epochs)...\n")

for epoch in range(51):
    # --- 1. TRAINING MODE ---
    net.train() # Turns ON Dropout and BatchNorm tracking

    optimizer.zero_grad()
    out_train = net(X_train)
    loss_train = loss_fn(out_train, y_train)
    loss_train.backward()
    optimizer.step()

    # --- 2. VALIDATION MODE ---
    # Every 10 epochs, freeze the network and test it fairly
    if epoch % 10 == 0:
        net.eval() # Turns OFF Dropout so all neurons are active for testing

        with torch.no_grad():
            out_val = net(X_val)
            loss_val = loss_fn(out_val, y_val)

            # Calculate accuracy on the unseen validation data
            preds = torch.argmax(out_val, dim=1)
            correct = (preds == y_val).sum().item()
            acc_val = (correct / len(y_val)) * 100

        print(f"Epoch {epoch:2d} | Train Loss: {loss_train.item():.4f} | Val Loss: {loss_val.item():.4f} | Val Acc: {acc_val:.2f}%")

print("\nDay 7 Advanced Training Complete!")
```
