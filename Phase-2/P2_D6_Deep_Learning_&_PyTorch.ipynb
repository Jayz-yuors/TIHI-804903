```python
import numpy as np
import torch
import torch.nn as nn

# 1. Simulate the dataset exactly as the bootcamp asked
# 1000 samples of 5 inputs (sand, silt, clay, pH, organic matter)
np.random.seed(42)
X_numpy = np.random.rand(1000, 5).astype(np.float32)

# Randomly assign 4 outputs (soil types): Clay (0), Sand (1), Loam (2), Silt (3)
y_numpy = np.random.randint(0, 4, size=(1000,))

# 2. Push to GPU
dev = "cuda" if torch.cuda.is_available() else "cpu"
X = torch.tensor(X_numpy, dtype=torch.float32, device=dev)
y = torch.tensor(y_numpy, dtype=torch.long, device=dev)

print("Simulated dataset created and moved to GPU!")
```

```python
# 1. Build the network from the Day 6 starter code
net = nn.Sequential(
    nn.Linear(5, 32), nn.ReLU(),
    nn.Linear(32, 16), nn.ReLU(),
    nn.Linear(16, 4)
).to(dev)

# 2. Define the loss (Softmax is built-in) and the optimizer
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(net.parameters(), lr=0.01)

print("Training the MLP...\n")

for epoch in range(101):
    # Forward pass: get the raw scores from the network
    raw_scores = net(X)

    # Calculate the error
    loss = loss_fn(raw_scores, y)

    # Backward pass: calculate gradients and update weights
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    # Track and print accuracy
    if epoch % 20 == 0:
        # The predicted class is the one with the highest score
        predictions = torch.argmax(raw_scores, dim=1)
        correct = (predictions == y).sum().item()
        accuracy = (correct / len(y)) * 100
        print(f"Epoch {epoch:3d} | Loss: {loss.item():.4f} | Accuracy: {accuracy:.2f}%")
```
